1. Using a simple example (Burglary–Alarm), create a Bayesian Network using

```python
# Bayesian Network using pgmpy
# Example: Burglary-Alarm Network

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
```

```python
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])
```

```python
# Step 2: Define Conditional Probability Tables

# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)
# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)
# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],  # Alarm = False
        [0.001, 0.29, 0.94, 0.95]   # Alarm = True
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)
# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],    # JohnCalls = False
        [0.05, 0.90]     # JohnCalls = True
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
    variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],    # MaryCalls = False
        [0.01, 0.70]     # MaryCalls = True
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
```

```python
# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)
```

```python
# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
```

```python
# Step 5: Perform inference
inference = VariableElimination(model)
# Query: Probability of Burglary when John and Mary both call
result = inference.query(
    variables=["Burglary"],
    evidence={
        "JohnCalls": "True",
        "MaryCalls": "True"
    }
)

print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")
print(result)
print("Siddhesh K(T009)")
```

```python
# Posterior Probability using Bayesian Network
# Query: P(Burglary | Alarm = True)

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
```

```python
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm")
])
```

```python
# Step 2: Define probability of Burglary
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)
```

```python
# Step 3: Define probability of Earthquake
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)
```

```python
# Step 4: Define probability of Alarm
# P(Alarm | Burglary, Earthquake)

cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],  # Alarm = False
        [0.001, 0.29, 0.94, 0.95]   # Alarm = True
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)
```

```python
# Step 5: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm
)
```

```python
# Step 6: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())
```

```python
# Step 7: Perform inference
inference = VariableElimination(model)

result = inference.query(
    variables=["Burglary"],
    evidence={"Alarm": "True"}
)

print("\nPosterior Probability:")
print("P(Burglary | Alarm = True)")
print(result)
print("Siddhesh K(T009)")
```

6A — Titanic Passenger Survival: Create Bayesian Network

```python
# Extra Task 6A
# Real-World Example: Hospital Patient Disease Diagnosis
# Bayesian Network using pgmpy

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
```

```python
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Smoking", "LungDisease"),
    ("Age", "LungDisease"),
    ("LungDisease", "Cough"),
    ("LungDisease", "BreathingProblem"),
    ("LungDisease", "ChestPain")
])
```

```python
# Step 2: Define Conditional Probability Tables

# P(Smoking)
cpd_smoking = TabularCPD(
    variable="Smoking",
    variable_card=2,
    values=[[0.70], [0.30]],
    state_names={"Smoking": ["False", "True"]}
)
# P(Age)
cpd_age = TabularCPD(
    variable="Age",
    variable_card=2,
    values=[[0.60], [0.40]],
    state_names={"Age": ["Young", "Old"]}
)
# P(LungDisease | Smoking, Age)
cpd_lung = TabularCPD(
    variable="LungDisease",
    variable_card=2,
    values=[
        [0.95, 0.80, 0.70, 0.40],  # LungDisease = False
        [0.05, 0.20, 0.30, 0.60]   # LungDisease = True
    ],
    evidence=["Smoking", "Age"],
    evidence_card=[2, 2],
    state_names={
        "LungDisease": ["False", "True"],
        "Smoking": ["False", "True"],
        "Age": ["Young", "Old"]
    }
)
# P(Cough | LungDisease)
cpd_cough = TabularCPD(
    variable="Cough",
    variable_card=2,
    values=[
        [0.80, 0.20],    # Cough = False
        [0.20, 0.80]     # Cough = True
    ],
    evidence=["LungDisease"],
    evidence_card=[2],
    state_names={
        "Cough": ["False", "True"],
        "LungDisease": ["False", "True"]
    }
)
# P(BreathingProblem | LungDisease)
cpd_breathing = TabularCPD(
    variable="BreathingProblem",
    variable_card=2,
    values=[
        [0.90, 0.30],    # BreathingProblem = False
        [0.10, 0.70]     # BreathingProblem = True
    ],
    evidence=["LungDisease"],
    evidence_card=[2],
    state_names={
        "BreathingProblem": ["False", "True"],
        "LungDisease": ["False", "True"]
    }
)
# P(ChestPain | LungDisease)
cpd_chestpain = TabularCPD(
    variable="ChestPain",
    variable_card=2,
    values=[
        [0.85, 0.40],    # ChestPain = False
        [0.15, 0.60]     # ChestPain = True
    ],
    evidence=["LungDisease"],
    evidence_card=[2],
    state_names={
        "ChestPain": ["False", "True"],
        "LungDisease": ["False", "True"]
    }
)
```

```python
# Step 3: Add CPDs to model
model.add_cpds(
    cpd_smoking,
    cpd_age,
    cpd_lung,
    cpd_cough,
    cpd_breathing,
    cpd_chestpain
)
```

```python
# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
```

```python
# Step 5: Display the Bayesian Network
print("\nBayesian Network Structure:")
print(model.edges())
```

```python
# Step 6: Perform inference
inference = VariableElimination(model)
# Query: Probability of Lung Disease
# when patient has cough, breathing problem and chest pain
result = inference.query(
    variables=["LungDisease"],
    evidence={
        "Cough": "True",
        "BreathingProblem": "True",
        "ChestPain": "True"
    }
)
print("\nProbability of Lung Disease given:")
print("Cough = True")
print("Breathing Problem = True")
print("Chest Pain = True")
print(result)
```

6B - Perform inference to calculate a posterior probability for a given observation.

```python
# Extra Task 6B
# Real-World Example: Loan Approval
# Query: P(LoanApproved | GoodCredit = True)

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
```

```python
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("GoodCredit", "LoanApproved"),
    ("HighIncome", "LoanApproved")
])
```

```python
# Step 2: Define probability of Good Credit
cpd_credit = TabularCPD(
    variable="GoodCredit",
    variable_card=2,
    values=[[0.30], [0.70]],
    state_names={"GoodCredit": ["False", "True"]}
)
```

```python
# Step 3: Define probability of High Income
cpd_income = TabularCPD(
    variable="HighIncome",
    variable_card=2,
    values=[[0.60], [0.40]],
    state_names={"HighIncome": ["False", "True"]}
)
```

```python
# Step 4: Define probability of Loan Approval
# P(LoanApproved | GoodCredit, HighIncome)

cpd_loan = TabularCPD(
    variable="LoanApproved",
    variable_card=2,
    values=[
        [0.95, 0.70, 0.60, 0.10],  # LoanApproved = False
        [0.05, 0.30, 0.40, 0.90]   # LoanApproved = True
    ],
    evidence=["GoodCredit", "HighIncome"],
    evidence_card=[2, 2],
    state_names={
        "LoanApproved": ["False", "True"],
        "GoodCredit": ["False", "True"],
        "HighIncome": ["False", "True"]
    }
)
```

```python
# Step 5: Add CPDs to model
model.add_cpds(
    cpd_credit,
    cpd_income,
    cpd_loan
)
```

```python
# Step 6: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())
```

```python
# Step 7: Perform inference
inference = VariableElimination(model)
result = inference.query(
    variables=["LoanApproved"],
    evidence={"GoodCredit": "True"}
)
```

```python
# Step 8: Display posterior probability
print("\nPosterior Probability:")
print("P(LoanApproved | GoodCredit = True)")
print(result)
```
